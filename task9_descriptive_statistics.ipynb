This notebook calculates descriptive statistics for a synthetic Titanic-style practice dataset. It is not the original Titanic dataset.

## 1. Import libraries and load data

```python
import pandas as pd
from pathlib import Path

# If running from this notebook in the project folder:
data_path = Path('../data/titanic_style_practice_dataset.csv')
df = pd.read_csv(data_path)
df.head()
```

## 2. Inspect the dataset

```python
print('Shape:', df.shape)
print('Data types:\n', df.dtypes)
print('Missing values:\n', df.isna().sum())
df.describe()
```

## 3. Calculate mean, median, mode, standard deviation, and percentiles

```python
columns = ['Age', 'Fare', 'SibSp', 'Parch', 'Pclass']
rows = []
for col in columns:
    s = df[col].dropna()
    modes = s.mode().tolist()
    rows.append({
        'Column': col,
        'Count': s.count(),
        'Mean': s.mean(),
        'Median': s.median(),
        'Mode': ', '.join(map(str, modes[:4])),
        'Std Dev (sample)': s.std(),
        'Min': s.min(),
        'Q1 (25%)': s.quantile(0.25),
        'Q2 (50%)': s.quantile(0.50),
        'Q3 (75%)': s.quantile(0.75),
        'Max': s.max(),
        'IQR': s.quantile(0.75) - s.quantile(0.25)
    })
summary = pd.DataFrame(rows)
summary.round(2)
```

## 4. Compare mean and median to inspect possible skew

```python
skew_check = summary[['Column', 'Mean', 'Median']].copy()
skew_check['Mean minus median'] = skew_check['Mean'] - skew_check['Median']
skew_check.round(2)
```

## 5. Export the summary table

```python
output_path = Path('../reports/summary_statistics.csv')
output_path.parent.mkdir(parents=True, exist_ok=True)
summary.to_csv(output_path, index=False)
print('Saved:', output_path)
```

## 6. Interpretation

- A mean greater than the median can indicate right skew; a mean lower than the median can indicate left skew. This is a clue, not a definitive skewness test.
- Standard deviation describes spread around the mean and is sensitive to extreme values.
- Q1 and Q3 bracket the middle 50% of observations; IQR = Q3 − Q1.
- Fare is expected to be right-skewed in this practice sample because fares were generated with a right-tailed distribution.
- Pclass is a coded category, so interpret its mean cautiously and also inspect its mode/frequency.

**Important:** The dataset is synthetic and should not be described as original Titanic data.
